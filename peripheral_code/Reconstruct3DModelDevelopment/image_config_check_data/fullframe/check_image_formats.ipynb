```python
import cv2            
import glob
```

```python
sink_video_path = glob.glob('*_sink.mp4')[0]        
source_video_path = glob.glob('*_source.mp4')[0]
```

```python
sink = cv2.VideoCapture(sink_video_path)
source = cv2.VideoCapture(source_video_path)
ret, source_frame = source.read()
ret, sink_frame = sink.read()
source_frame = cv2.flip(source_frame, -1)
sink_frame = cv2.flip(sink_frame, -1)
print(source_frame.shape)
cv2.imwrite('rpicam_full_frame_sink.jpeg', sink_frame)
cv2.imwrite('rpicam_full_frame_source.jpeg', source_frame)

sink.release()
source.release()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib auto
import matplotlib
matplotlib.rcParams.update({'font.size': 20})
import glob
import os
```

```python
data = np.load('./stereo_A1_fullframe.npy', allow_pickle=True).item()
Kl, Dl, Kr, Dr, R, T, img_size = data['Kl'], data['Dl'], data['Kr'], data['Dr'], \
                                 data['R'], data['T'], data['img_size']
```

```python
R1, R2, P1, P2, Q, validRoi1, validRoi2 = cv2.stereoRectify(Kl, Dl, Kr, Dr, img_size, R, T)
```

```python
xmap1, ymap1 = cv2.initUndistortRectifyMap(Kl, Dl, R1, P1, img_size, cv2.CV_32FC1)
xmap2, ymap2 = cv2.initUndistortRectifyMap(Kr, Dr, R2, P2, img_size, cv2.CV_32FC1)
```

```python
dest = r"calibrated_and_rectified"
left_image = './rpicam_full_frame_source.jpeg'
right_image = './rpicam_full_frame_sink.jpeg'



img_l = cv2.imread(left_image)
img_r = cv2.imread(right_image)
left_img_rectified = cv2.remap(img_l, xmap1, ymap1, cv2.INTER_LINEAR)
right_img_rectified = cv2.remap(img_r, xmap2, ymap2, cv2.INTER_LINEAR)

cv2.imshow("undistored_and_rectified_left", left_img_rectified)
cv2.imshow("undistored_and_rectified_right", right_img_rectified)
print(left_image, right_image)

cv2.imwrite(left_image,left_img_rectified)
cv2.imwrite(right_image, right_img_rectified)
cv2.waitKey(0)
    
cv2.destroyAllWindows()
```
