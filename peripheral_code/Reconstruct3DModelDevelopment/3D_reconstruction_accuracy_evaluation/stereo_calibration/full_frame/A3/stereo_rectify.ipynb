```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib auto
import matplotlib
matplotlib.rcParams.update({'font.size': 20})
import glob
import os             
```

```python
data = np.load('./stereo.npy', allow_pickle=True).item()
Kl, Dl, Kr, Dr, R, T, img_size = data['Kl'], data['Dl'], data['Kr'], data['Dr'], \
                                 data['R'], data['T'], data['img_size']
```

```python
Kl, Kr
```

```python
R1, R2, P1, P2, Q, validRoi1, validRoi2 = cv2.stereoRectify(Kl, Dl, Kr, Dr, img_size, R, T)
```

```python
xmap1, ymap1 = cv2.initUndistortRectifyMap(Kl, Dl, R1, P1, img_size, cv2.CV_32FC1)
xmap2, ymap2 = cv2.initUndistortRectifyMap(Kr, Dr, R2, P2, img_size, cv2.CV_32FC1)
```

```python
dest = r"calibrated_and_rectified"
left_images = glob.glob('./calibration_images/*left.jpeg')
right_images = glob.glob('./calibration_images/*right.jpeg')

for left, right in zip(left_images, right_images):

    left_filename = left.split("\\")[-1].replace(".jpeg", "_undistored_and_rectified.jpeg")
    right_filename = right.split("\\")[-1].replace(".jpeg", "_undistored_and_rectified.jpeg")

    img_l = cv2.imread(left)
    img_r = cv2.imread(right)
    left_img_rectified = cv2.remap(img_l, xmap1, ymap1, cv2.INTER_LINEAR)
    right_img_rectified = cv2.remap(img_r, xmap2, ymap2, cv2.INTER_LINEAR)

    cv2.imshow("undistored_and_rectified_left", left_img_rectified)
    cv2.imshow("undistored_and_rectified_right", right_img_rectified)
    print(left_filename, right_filename)
    cv2.imwrite(os.path.join(dest, left_filename),left_img_rectified)
    cv2.imwrite(os.path.join(dest, right_filename),         right_img_rectified)
    cv2.waitKey(0)
    
cv2.destroyAllWindows()
```
