```python
import cv2
```

```python
frame = cv2.imread('left_rectified.png',0)
res, corners = cv2.findChessboardCorners(frame, (9,6))

x1, y1 = int(corners[0][0][0]), int(corners[0][0][1]) # top left
x2, y2 = int(corners[8][0][0]), int(corners[8][0][1]) # top right
x3, y3 = int(corners[-9][0][0]), int(corners[-9][0][1]) # bottom left
x4, y4 = int(corners[-1][0][0]), int(corners[-1][0][1]) # bottom right


template = frame[y1: y4, x1:x4]
target_scanline = frame[y1:y4,:]

cv2.drawChessboardCorners(frame, (9,6), corners, res)
# cv2.circle(frame, (x4, y4), 4, (255, 255, 255), 2, 2)
cv2.imshow("frame", frame)
cv2.imshow("template", template)
cv2.imshow('target_scanline', target_scanline)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# read input and template image
img = cv2.imread('right_rectified.png')
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
target_scanline = img_gray[y1:y4,:]

w, h = template.shape[::-1]
 
# Apply template Matching
res = cv2.matchTemplate(target_scanline,template,cv2.TM_SQDIFF)

# Find min/max value and location
min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(res)
 
# If the method is TM_SQDIFF or TM_SQDIFF_NORMED, 
# take minimum otherwise maximum
bottom_right = (min_loc[0] + w, min_loc[1] + h)
 
# draw the rectangle
cv2.rectangle(img,(min_loc[0], y1), (bottom_right[0], y4), (0,0,255), 2)
 
# Display the image
cv2.imshow('a', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
