```python
import cv2
import os
import glob
```

```python
cap_left = cv2.VideoCapture(glob.glob(r".\calib_videos\*_left.avi")[0])
cap_right = cv2.VideoCapture(glob.glob(r".\calib_videos\*_right.avi")[0])
dest = r'stereo_calib_data/ittr5'
save = True
count = 0
while True:
    ret1, left_frame = cap_left.read()
    ret2, right_frame = cap_right.read()

    if not ret1 and not ret2:
        break
    cv2.imshow("left_frame", left_frame)
    cv2.imshow("right_frame", right_frame)

    cv2.waitKey(10)
    if save:    
        path_left = os.path.join(dest, f"frame_{count}_left.jpeg")
        cv2.imwrite(path_left, left_frame)

        path_right = os.path.join(dest, f"frame_{count}_right.jpeg")
        cv2.imwrite(path_right, right_frame)
    count += 1

cv2.destroyAllWindows()
cap_right.release()
cap_left.release()
```
