```python
import cv2            
import glob 
```

```python
sink_video_path = glob.glob('fullframe_data/*sink.mp4')[0]           
source_video_path = glob.glob('fullframe_data/*source.mp4')[0]
```

```python
sink = cv2.VideoCapture(sink_video_path)
source = cv2.VideoCapture(source_video_path)
count = 0

while True:
    ret, source_frame = source.read()
    ret, sink_frame = sink.read()
    if not ret:
        break
    count += 1
    cv2.imshow("source", cv2.flip(source_frame, -1))
    cv2.imshow("sink", cv2.flip(sink_frame, -1))
    cv2.waitKey(10)                     
count
cv2.destroyAllWindows()
print(count)
sink.release()
source.release()
```

```python
import pandas as pd                                                   
```

```python
source = pd.read_csv(source_video_path.replace(".mp4", ".txt"), header=None, names=['timestamp'])
sink = pd.read_csv(sink_video_path.replace(".mp4", ".txt"), header=None, names=['timestamp'])
```

```python
sink.head()
```

```python
source.timestamp = source.timestamp.apply(lambda t: float((t.split("=")[-1])))
sink.timestamp = sink.timestamp.apply(lambda t: float((t.split("=")[-1])))
```

```python
df = pd.concat([source, sink], axis=1)
df.columns =['source_ts', 'sink_ts']
df['delta'] = df['sink_ts'] - df['source_ts']
```
