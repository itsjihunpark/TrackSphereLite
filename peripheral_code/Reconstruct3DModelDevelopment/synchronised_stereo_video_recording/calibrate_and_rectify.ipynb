```python
import cv2
import numpy as np
import glob
import os
```

```python
data = np.load('./stereo_A1_fullframe.npy', allow_pickle=True).item()
Kl, Dl, Kr, Dr, R, T, img_size = data['Kl'], data['Dl'], data['Kr'], data['Dr'], \
                                 data['R'], data['T'], data['img_size']
```

```python
R1, R2, P1, P2, Q, validRoi1, validRoi2 = cv2.stereoRectify(Kl, Dl, Kr, Dr, img_size, R, T)
```

```python
xmap1, ymap1 = cv2.initUndistortRectifyMap(Kl, Dl, R1, P1, img_size, cv2.CV_32FC1)
xmap2, ymap2 = cv2.initUndistortRectifyMap(Kr, Dr, R2, P2, img_size, cv2.CV_32FC1)
```

```python
dest = r"calibrated_and_rectified"
right_image = './frames/12.png'


img_r = cv2.imread(right_image)
right_img_rectified = cv2.remap(img_r, xmap2, ymap2, cv2.INTER_LINEAR)
cv2.imshow("undistored_and_rectified_right", right_img_rectified)
cv2.imwrite('undistored_and_rectified_right.png', right_img_rectified)
cv2.waitKey(0)
    
cv2.destroyAllWindows()
```
