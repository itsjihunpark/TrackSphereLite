```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
def plot_results(loss_values_train, loss_values_val, epoch, type_of_model, loss_type):
    plt.figure(figsize=(8,6), dpi=75)
    plt.plot(epoch,loss_values_train, label='train')
    plt.plot(epoch, loss_values_val, label='val')
    plt.xlabel("epoch")
    plt.ylabel(loss_type)
    plt.title(type_of_model)
    plt.legend()
```

```python
df_fine_tuned = pd.read_csv("fine_tuned_results.csv")
df_from_scratch = pd.read_csv("from_scratch_results.csv")
```

```python
plot_results(df_fine_tuned['train/box_loss'], df_fine_tuned['val/box_loss'],df_fine_tuned['epoch'], "Fine tuned model; bounding box loss over 100 epoch", "bouding box loss")
```

```python
plot_results(df_from_scratch['train/box_loss'], df_from_scratch['val/box_loss'],df_from_scratch['epoch'], "trained from scratch; bounding box loss over 100 epoch", "bouding box loss")
```

```python
plot_results(df_fine_tuned['train/cls_loss'], df_fine_tuned['val/cls_loss'],df_fine_tuned['epoch'], "Fine tuned model; classification loss over 100 epoch", "classification loss")
```

```python
plot_results(df_from_scratch['train/cls_loss'], df_from_scratch['val/cls_loss'],df_from_scratch['epoch'], "trained from scratch; classification loss over 100 epoch", "classification loss")
```
