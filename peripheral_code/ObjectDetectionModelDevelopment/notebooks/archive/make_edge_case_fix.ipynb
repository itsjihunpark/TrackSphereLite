Edge case: unintended placement of partially visible golf ball which can be seen on every PR_Library_Top_Floor left frame
See example.
This notebook will crop out this area


```python
import cv2
import os
import glob
```

```python
Root_Data_Dir = ""
left_video_dir = glob.glob(os.path.join(Root_Data_Dir,"*_source.mp4"))
```

```python
for dir in left_video_dir:
    frames_path = os.listdir(dir)
    for frame in frames_path:
        frame_path = os.path.join(dir,frame)
        img = cv2.imread(frame_path)
        cropped_img = img[:,70:]
        cv2.imwrite(frame_path, cropped_img)
```
