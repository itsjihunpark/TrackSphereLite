```python
import os
import shutil
```

```python
def move_file(source,destination):

    # Move the file
    shutil.move(source, destination)

    print(f"File moved to {destination}")
```

```python
def make_dir(folder_path):
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)
        print(f"Folder created at: {folder_path}")
    else:
        print(f"Folder already exists at: {folder_path}")
```

```python
ROOT_DIR = "../data/Processing/Golf_Ball_Flighted_data"
with open("pre-labelling.csv","r") as f:
    for line in f.readlines():
        line = line.strip().split(",")
        start, end = line[2], line[3]
        if start == "NA":
            continue
        video_path = os.path.join(ROOT_DIR,line[0], "None_Strike_frames",line[1])


        for i, frame in enumerate(os.listdir(video_path)):
            
            video_name = video_path.split("\\")[-1]
            filename = f"{video_name}_frame_right_{i}.jpeg"
            src_frame_path = os.path.join(video_path, filename)
            if i>=int(start) and i<=int(end):
                if os.path.exists(src_frame_path):
                    print(src_frame_path)
                
```
