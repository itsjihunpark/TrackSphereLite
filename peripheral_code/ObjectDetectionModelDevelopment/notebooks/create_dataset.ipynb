## Actually copying over the files to base_dataset folder

```python
from shutil import copy
import os
import pandas as pd
```

```python
Root_Dataset_Folder = "../data/training_data/baseline_dataset"
ROOT_original_dataset = "../data/labelled_golf_ball_data"
```

```python
def make_dir(folder_path):
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)
        print(f"Folder created at: {folder_path}")
    else:
        print(f"Folder already exists at: {folder_path}")
```

```python
splitted_set = pd.read_csv("baseline_dataset.csv")
```

## Creating baseline dataset train(80)-val(10)-test(10)

```python
baseline_dataset = {}
baseline_dataset['train'] = splitted_set[splitted_set['train_set']==1]
baseline_dataset['val'] = splitted_set[splitted_set['val_set']==1]
baseline_dataset['test'] = splitted_set[splitted_set['test_set']==1]
```

```python
for key in baseline_dataset.keys():    
    images_folder_path = os.path.join(Root_Dataset_Folder, "images", key)
    labels_folder_path = os.path.join(Root_Dataset_Folder, "labels", key)
    
    make_dir(images_folder_path)
    make_dir(labels_folder_path)
   
    
    for i, row in baseline_dataset[key].iterrows():
        copy(os.path.join(ROOT_original_dataset, row['frame_path']), images_folder_path)
        copy(os.path.join(ROOT_original_dataset, row['label_path']), labels_folder_path)
```
