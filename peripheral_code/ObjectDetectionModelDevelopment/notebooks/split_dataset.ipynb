```python
import pandas as pd
import os
import glob
```

## Go through dataset folder ("../data/labelled_golf_ball_data") and produce base_dataset.csv

```python
ROOT_original_dataset = "../data/labelled_golf_ball_data"
count = 0
with open("base_dataset.csv", "w") as f:
    csv_header = f"location,swing_type,frame_type,src_video,frame_path,label_path\n"
    f.write(csv_header)
    for dir in os.listdir(ROOT_original_dataset):
        golf_ball_labelled_dir = os.path.join(ROOT_original_dataset,dir)

        for location in os.listdir(golf_ball_labelled_dir):
            
            location_dir = os.path.join(golf_ball_labelled_dir,location)
            
            for frame_type in os.listdir(location_dir):
                video_dir = os.path.join(location_dir,frame_type)
                for src_video in os.listdir(video_dir):
                    frames_dir = os.path.join(video_dir, src_video, "images")
                    labels_dir = os.path.join(video_dir, src_video, "labels")
                    for frame, label in zip(os.listdir(frames_dir),os.listdir(labels_dir)):
                        
                        frame = f"images\{frame}"
                        label = f"labels\{label}"
                        print(f"dir: {dir}")
                        print(f"    location: {location}")
                        print(f"        frame_type: {frame_type}")
                        print(f"            src_video: {src_video}")
                        print(f"                frame: {frame}")
                        print(f"                frame: {label}")
                        
                        dir
                        location
                        swing_type = dir.replace("Golf_Ball_","").replace("_data_labelled_final","")
                        frame_type
                        src_video
                        frame_path = os.path.join(dir, location, frame_type, src_video,frame)
                        label_path = os.path.join(dir, location, frame_type, src_video,label)
                        
                        csv_line = f"{location},{swing_type},{frame_type},{src_video},{frame_path},{label_path}\n"
                        f.write(csv_line)
                        count+=1
print(f"data count: {count}")
```

```python
df = pd.read_csv("./base_dataset.csv")
df['train_set'] = 0
df['val_set'] = 0
df['test_set'] = 0
df.head()
```

```python
grouped = df.groupby("swing_type")
```

```python
df_group = {}
for name, group in grouped:
    df_group[name] = group
```

```python
flighted_df_group = {}
for name, group in df_group['Flighted'].groupby("location"):
    print(name)
    flighted_df_group[name] = group
```

```python
putting_df_group = {}
for name, group in df_group['Putting'].groupby("location"):
    print(name)
    putting_df_group[name] = group
```

## Stratefied data splitting (train-val-test) for every location (where data was collected) and for every frame type (strike_frame/non_strike_frame)

```python
from sklearn.model_selection import train_test_split
```

```python
splitted_set = pd.DataFrame(columns=df.columns)
splitted_set.head()
```

```python
flighted_df_group['KP_Common_Area'].head(2)
```

```python
for key in flighted_df_group.keys():
    print(key)
    print(flighted_df_group[key].value_counts("frame_type"))
    print()
    for name, group in flighted_df_group[key].groupby("frame_type"):
        print(f"{name} total", group['frame_type'].count())
        # splitting here
        train, test = train_test_split(group, test_size=0.2, random_state=123)
        test, val = train_test_split(test, test_size=0.5, random_state=123)
        
        train['train_set'] = 1
        test['test_set'] = 1
        val['val_set'] = 1

        print("train", len(train))
        print('test', len(test))
        print('val', len(val))
        # re concat here
        splitted_set = pd.concat([splitted_set, train, test, val], ignore_index=True)


        
    print()
    print()
```

```python
for key in putting_df_group.keys():
    print(key)
    print(putting_df_group[key].value_counts("frame_type"))
    print()
    for name, group in putting_df_group[key].groupby("frame_type"):
        print(f"{name} total", group['frame_type'].count())
        # splitting here
        train, test = train_test_split(group, test_size=0.2, random_state=123)
        test, val = train_test_split(test, test_size=0.5, random_state=123)
        
        train['train_set'] = 1
        test['test_set'] = 1
        val['val_set'] = 1
        
        
        print("train", len(train))
        print('test', len(test))
        print('val', len(val))
        # re concat here
        splitted_set = pd.concat([splitted_set, train, val, test], ignore_index=True)
    print()
    print()
```

```python
print(f"train {splitted_set['train_set'].value_counts()[1]}")
print(f"val {splitted_set['val_set'].value_counts()[1]}")
print(f"test {splitted_set['test_set'].value_counts()[1]}")
```
